````python

This notebook walks through adding `deped_logo.png` as the application logo in the Laravel Blade layout.

## Steps
1. Verify the logo file and public asset path.
2. Add `deped_logo.png` to the Laravel public asset folder.
3. Update the Blade layout to render the logo.
4. Verify the logo displays from the public asset path.
5. Run a local server to preview the changes.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">import os
from pathlib import Path

project_root = Path(r"c:\Users\cmndr\myproject")
logo_path = project_root / "deped_logo.png"
public_logo_path = project_root / "public" / "deped_logo.png"
print("Project root:", project_root)
print("Logo exists:", logo_path.exists())
print("Public logo exists:", public_logo_path.exists())
print("Public path:", public_logo_path)
</VSCode.Cell>
<VSCode.Cell id="3" language="markdown">## Add `deped_logo.png` to Public Assets

If the logo is not in `public/`, copy it there.

Use a Python command below to copy the file from the project root into `public/deped_logo.png`.
</VSCode.Cell>
<VSCode.Cell id="4" language="python">import shutil

if logo_path.exists():
    public_dir = project_root / "public"
    public_dir.mkdir(parents=True, exist_ok=True)
    shutil.copy2(logo_path, public_logo_path)
    print(f"Copied {logo_path} to {public_logo_path}")
else:
    print(f"Source logo not found: {logo_path}")

print("Public logo exists after copy:", public_logo_path.exists())
</VSCode.Cell>
<VSCode.Cell id="5" language="markdown">## Update Blade Layout to Display the Logo

Edit `resources/views/layout/app.blade.php` and replace or add the sidebar brand HTML to render the logo using `asset('deped_logo.png')`.

The layout should use an `<img>` tag inside `.sidebar-brand`.
</VSCode.Cell>
<VSCode.Cell id="6" language="python">from pathlib import Path

layout_file = project_root / "resources" / "views" / "layout" / "app.blade.php"
layout_content = layout_file.read_text(encoding="utf-8")
old_block = "<div class=\"sidebar-brand\">\n        <i class=\"ti ti-users\" aria-hidden=\"true\"></i>\n        <span>HR Appointment</span>\n    </div>"
new_block = "<div class=\"sidebar-brand\">\n        <img src=\"{{ asset('deped_logo.png') }}\" alt=\"DepEd logo\" class=\"sidebar-logo\">\n        <span>HR Appointment</span>\n    </div>"

if old_block in layout_content:
    layout_file.write_text(layout_content.replace(old_block, new_block), encoding="utf-8")
    print("Updated sidebar brand block in app.blade.php")
else:
    print("Old block not found in layout; please update manually.")
</VSCode.Cell>
<VSCode.Cell id="7" language="markdown">## Render and Verify the Logo in the Notebook

Display the logo from the public path to verify it exists and can be rendered.
</VSCode.Cell>
<VSCode.Cell id="8" language="python">from IPython.display import Image, display

if public_logo_path.exists():
    display(Image(filename=str(public_logo_path)))
else:
    print("Public logo not found:", public_logo_path)
</VSCode.Cell>
<VSCode.Cell id="9" language="markdown">## Run Local Server and Preview Changes

Use the local Laravel development server to preview the app. The logo should appear in the sidebar brand area.

Run this command in a terminal:

```bash
php artisan serve --host=127.0.0.1 --port=8000
```

Then open `http://127.0.0.1:8000` in your browser.
````
